```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("/content/Live.csv")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True,axis=1)
```

```python
df.head()
```

```python
df.drop(columns=["status_id","status_published"],axis=1)
```

```python
#feature extraction-determining which value goes to the x and y axis(or target variable)
x=df
y=df["status_type"]
```

```python
#convert the target variable to numbers (using a label encoder)
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
x['status_type']=le.fit_transform(x['status_type'])
y=le.transform(y)#one column
```

```python
x.head()
```

```python
y[:5]
```

```python
#feature scaling (to improve the performance of the algorithm and prevent it from being biased)
x = x.drop(columns=["status_id", "status_published"])
cols=x.columns#x.columns has input variables that affect target variables(yaxis)
from sklearn.preprocessing import MinMaxScaler #scales data based on ranges
#initialize scaler
sc=MinMaxScaler()
x=sc.fit_transform(x)
#convert the data back into a dataframe
x=pd.DataFrame(x,columns=[cols])
```

```python
from sklearn.cluster import KMeans
km=KMeans(n_clusters=5,random_state=0)
```

```python
#train the algorithm
km.fit(x)
```

```python
#cluster centers
km.cluster_centers_#labels-help to identify each category
labels=km.labels_
#check the number of correct labels
correct_labels=sum(y==labels)
#check how accurate the algorithm is
print("Result: %d out of %d labels were correctly predicted." % (correct_labels,y.size))
```

```python
#check the model's accuracy
print("Accuracy:{0:0.2f}".format(correct_labels/float(y.size)))
```

```python
#elbow technique to show the optimum number of clusters
cs=[]#store the model's performance accross different clusters
for i in range(1,11):
  km=KMeans(n_clusters=i,init="k-means++",max_iter=350,n_init=10,random_state=0)
  km.fit(x)#train the algorithm
  #add the model's performance to a list
  cs.append(km.inertia_)
#plot a graph
plt.plot(range(1,11),cs)
plt.title("The Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("Model's performance")
plt.show()
```

```python
km=KMeans(n_clusters=1,random_state=0)
km.fit(x)
#cluster centers
km.cluster_centers_#labels-help to identify each category
labels=km.labels_
#check the number of correct labels
correct_labels=sum(y==labels)
#check how accurate the algorithm is
print("Result: %d out of %d labels were correctly predicted." % (correct_labels,y.size))
#check the model's accuracy
print("Accuracy:{0:0.2f}".format(correct_labels/float(y.size)))
```
